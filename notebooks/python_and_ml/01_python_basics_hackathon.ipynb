# 01 - Python Basics for Hackathons

This notebook introduces essential Python syntax, variable types, operators, collections, and functions for beginners and hackathon participants.

```python
print("Hello World")
```

```python
number1 = 12
number2 = 18

total = number1 + number2
print("Sum:", total)
```

```python
# Python is case sensitive
sales = 1000007
myName = "Shubham"
print("Name:", myName)
print("Sales:", sales)
```

```python
# Numeric data types
a = 124        # integer
b = 12.433     # float
c = 10 + 4j    # complex number (in Python, "j" denotes the imaginary unit)

print(f"a = {a} ({type(a).__name__})")
print(f"b = {b} ({type(b).__name__})")
print(f"c = {c} ({type(c).__name__})")
```

```python
# Boolean values in Python evaluate to 1 and 0
print("True == 1 :", True == 1)
print("False == 0:", False == 0)
```

```python
# Strings and multi-line formatting
name = "Shubham"
myName = 'Shubham'
bio = """My name is Shubham.
I am learning Machine Learning and Quantum Computing."""
print(bio)
```

```python
# Lists: ordered, mutable collections of items
my_list = [10, "Python", 3.14, True]
print("List:", my_list)
print("First element:", my_list[0])
```

```python
# Tuples: ordered, immutable collections
T1 = (0, 3, 36, 32, 35)
print("Tuple:", T1)
print("Length:", len(T1))
```

```python
# Dictionaries: key-value pairs
Myself = {
    "Name": "Shubham",
    "Topic": "Deep Learning",
    "Age": 18,
}
print("Dictionary:", Myself)
print("Topic:", Myself["Topic"])
```

```python
# Sets: unordered collections of unique elements
thisIsASet = {12, "shubh", "mahi", 12}
print("Set (duplicates automatically removed):", thisIsASet)
```

```python
# Arithmetic division
print("Float division (3 / 1):", 3 / 1)
print("Float division (5 / 2):", 5 / 2)
print("Floor division (5 // 2):", 5 // 2)
```

```python
# Input and type casting (with safe fallback for non-interactive execution)
myAge = "18"
userName = "Shubham"
userHeight = 5.9
print(f"User profile -> Name: {userName}, Age: {myAge}, Height: {userHeight}")
```

```python
# Number arithmetic
num1 = 15
num2 = 25
print("Sum:", num1 + num2)
```

```python
# Functions with default arguments
def myFunction(country="India"):
    print("I am from " + country)

myFunction("Sweden")
myFunction()
```
