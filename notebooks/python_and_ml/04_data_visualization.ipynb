# 04 - Data Visualization with Matplotlib & Pandas

Visualizes patterns in the vehicle observation dataset (`data/data_collection.csv`), including top observation locations and record frequencies over time.

```python
import os
import pandas as pd
import matplotlib.pyplot as plt

# Load dataset
csv_path = os.path.join("..", "..", "data", "data_collection.csv")
if not os.path.exists(csv_path):
    csv_path = "data/data_collection.csv"

df = pd.read_csv(csv_path)
df.head()
```

### 1. Top 10 Locations with Highest Vehicle Sightings

```python
plt.figure(figsize=(10, 5))
top_locations = df["Location"].value_counts().head(10)
top_locations.plot(kind="bar", color="#2b5c8f", edgecolor="black")
plt.title("Top 10 Locations by Record Count", fontsize=14)
plt.xlabel("City / Location", fontsize=12)
plt.ylabel("Number of Sightings", fontsize=12)
plt.xticks(rotation=45, ha="right")
plt.grid(axis="y", linestyle="--", alpha=0.7)
plt.tight_layout()
plt.show()
```

### 2. Sightings Trend Across Dates

```python
plt.figure(figsize=(12, 4))
date_counts = df["Date"].value_counts().sort_index()
date_counts.plot(kind="line", marker="o", color="#e76f51", linewidth=2)
plt.title("Daily Vehicle Observations Trend", fontsize=14)
plt.xlabel("Date", fontsize=12)
plt.ylabel("Observations", fontsize=12)
plt.xticks(rotation=45, ha="right")
plt.grid(True, linestyle="--", alpha=0.5)
plt.tight_layout()
plt.show()
```
