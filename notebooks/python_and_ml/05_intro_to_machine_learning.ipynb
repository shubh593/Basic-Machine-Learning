## The Machine Learning Lifecycle

1. **Data Gathering**: Collecting raw domain data from sensors, databases, or APIs.
2. **Data Preparation**: Cleaning, handling missing values, encoding categoricals, and feature scaling.
3. **Model Selection**: Choosing an appropriate algorithm (e.g., Linear Regression, Random Forest, Neural Network).
4. **Training**: Feeding training features to the model to learn weights/parameters.
5. **Evaluation**: Measuring accuracy, loss, precision, recall, or RMSE on test/validation data.
6. **Hyperparameter Tuning**: Optimizing configuration parameters (learning rate, tree depth).
7. **Prediction & Deployment**: Running inference on unseen data.

## Learning Paradigms

- **Supervised Learning**: Model learns from labeled input-output pairs $(X, y)$.
  - *Regression*: Predicting continuous numbers (e.g., house prices, sales, temperature).
  - *Classification*: Predicting discrete categories (e.g., spam vs ham, disease diagnosis).
- **Unsupervised Learning**: Model discovers patterns or clusters in unlabeled data $(X)$.
  - *Clustering*: Grouping similar instances (e.g., customer segmentation).
  - *Dimensionality Reduction*: Compressing features (e.g., PCA, t-SNE).
- **Reinforcement Learning**: Agent learns optimal actions through environment rewards and penalties.

## Minimal Working Example: Linear Regression in Scikit-Learn

Below is a complete, minimal implementation of a Supervised Linear Regression model predicting exam score based on study hours.

```python
import numpy as np
from sklearn.linear_model import LinearRegression

# 1. Training features (Hours studied) and target (Exam score)
X = np.array([[1], [2], [3], [4], [5]])
y = np.array([50, 60, 70, 80, 90])

# 2. Instantiate and train model
model = LinearRegression()
model.fit(X, y)

# 3. Predict for new values
test_hours = np.array([[6], [7.5]])
predictions = model.predict(test_hours)

for hours, score in zip(test_hours.ravel(), predictions):
    print(f"Hours studied: {hours:4.1f} -> Predicted score: {score:.1f}")

print(f"\nModel Formula: Score = {model.coef_[0]:.1f} * Hours + {model.intercept_:.1f}")
```
