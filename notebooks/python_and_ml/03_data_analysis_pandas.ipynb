```python
# List slicing
myList = [20, 30, 40, 50, 70]
print("Sliced elements:", myList[3:])
```

```python
# Dictionaries
trainer_dict = {"raghu": "trainer", "mahi": "player", "guru": "mentor"}
print("Mahi role:", trainer_dict["mahi"])
print("Dictionary keys:", list(trainer_dict.keys()))
```

```python
# Nested structures
students = {"s_1": ["IBM"], "s_2": ["infy", "wipro", "google"]}
print("Third company for student 2:", students["s_2"][2])
```

```python
# Lists vs Tuples (mutability)
tup_ex = (1, 2, 3, 4, 4, 2)
mylist = [1, 2, 3, 4, 5]
mylist.append(7)
print("Updated mutable list:", mylist)
print("Immutable tuple:", tup_ex)
```

## Loading and Exploring Data with Pandas

```python
import os
import pandas as pd

# Resolve relative path to dataset
csv_path = os.path.join("..", "..", "data", "data_collection.csv")
if not os.path.exists(csv_path):
    csv_path = "data/data_collection.csv"

df = pd.read_csv(csv_path)
print("Dataset shape:", df.shape)
df.head(10)
```

```python
print("Tail records:")
df.tail()
```

```python
# Inspect columns and Location distribution
print("Columns in dataset:", df.columns.tolist())
print("\nTop 5 Most Frequent Locations:")
print(df["Location"].value_counts().head(5))
```

```python
# Working with temporary DataFrames and type casting
temp = pd.DataFrame({
    "A": [1, 2, 3],
    "B": [11, 22, 33],
    "C": ["12-03-2024", "13-04-2024", "13-05-2024"]
})
print("Before casting:")
print(temp.dtypes)
temp
```

```python
temp["A"] = pd.to_numeric(temp["A"])
temp["C"] = pd.to_datetime(temp["C"], format="%d-%m-%Y")
print("After casting:")
print(temp.dtypes)
temp
```
