# 06 - Grover's Algorithm & Oracle Basics

An introduction to quantum search, phase oracles using controlled-Z (`cz`), circuit composition, and statevector visualization.

```python
my_list = [39, 61, 17, 12]
```

```python
# Classical oracle definition
def the_oracle(my_input):
    winner = 17
    return my_input == winner

print("Oracle test (17):", the_oracle(17))
print("Oracle test (39):", the_oracle(39))
```

```python
import math
import numpy as np
import matplotlib.pyplot as plt
from qiskit import QuantumCircuit
from qiskit.circuit.library import GroverOperator, MCMT, ZGate
from qiskit.visualization import plot_distribution
```

```python
# Build 2-qubit phase oracle marking state |11>
oracle = QuantumCircuit(2, name="oracle")
oracle.cz(0, 1)
oracle.draw("mpl")
```

```python
# Assemble Grover search circuit
grover_circ = QuantumCircuit(2, 2)
grover_circ.h([0, 1])  # Equal superposition
grover_circ.append(oracle.to_gate(), [0, 1])  # Apply oracle
grover_circ.draw("mpl")
```

```python
# Visualize quantum state on Bloch spheres
from qiskit.quantum_info import Statevector
from qiskit.visualization import plot_bloch_multivector

state = Statevector.from_instruction(grover_circ)
plot_bloch_multivector(state)
```
