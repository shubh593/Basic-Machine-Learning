```python
from qiskit.circuit.library import RealAmplitudes
from qiskit.circuit import QuantumCircuit, QuantumRegister, ClassicalRegister
from qiskit.quantum_info import SparsePauliOp
from qiskit.transpiler.preset_passmanagers import generate_preset_pass_manager
from qiskit_ibm_runtime.fake_provider import FakeManilaV2
from qiskit_ibm_runtime import SamplerV2 as Sampler

# Bell Circuit
qc = QuantumCircuit(2)
qc.h(0)
qc.cx(0, 1)
qc.measure_all()

# Run sampler job locally using FakeManilaV2 mock backend
fake_manila = FakeManilaV2()
pm = generate_preset_pass_manager(backend=fake_manila, optimization_level=1)
isa_qc = pm.run(qc)

# Fix seed for reproducible results
options = {"simulator": {"seed_simulator": 42}}
sampler = Sampler(mode=fake_manila, options=options)

result = sampler.run([isa_qc]).result()
print("Local FakeManilaV2 Sampler result:", result)
```

```python
from qiskit import QuantumCircuit
from qiskit.visualization import plot_histogram
from qiskit_aer import AerSimulator

# 1. Create circuit with 1 qubit and 1 classical bit
qc = QuantumCircuit(1, 1)
qc.h(0)  # Put qubit 0 into superposition (|0> + |1>) / sqrt(2)
qc.measure(0, 0)

# 2. Simulate with AerSimulator
backend = AerSimulator()
result = backend.run(qc).result()
counts = result.get_counts()
print("Superposition Measurement Counts:", counts)
plot_histogram(counts)
```
