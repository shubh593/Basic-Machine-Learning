```python
df = df.drop(['Category'],axis=1)
df.head()
```

                   By Mean

```python
# putting null values by mean of it
df['Price']=df['Price'].fillna(df['Price'].mean())
df['Rating']=df['Rating'].fillna(df['Rating'].mean())
```

```python
df.head()
```

```python
# Lets assume price over rating
x=df['Rating']
y=df['Price']
```

```python
# By defining B1 and B0
x_mean=df['Price'].mean()
y_mean=df['Rating'].mean()
print(x_mean)
print(y_mean)
```

```python
B1=((x-x_mean)*(y-y_mean)).sum()/((x-x_mean)**2).sum()
B0=y_mean-B1*x_mean
print(B1)
print(B0)
```

```python
y_predicted=B0 + B1*x
```

```python
y_predicted
```

```python
df.head()
```

```python
error = y - y_predicted
error
```

```python
mean_squared_error = (error**2).mean()
print(np.sqrt(mean_squared_error))
```

**                    By Mode**

```python

# 1. Calculate the mode as an integer scalar
price_mode = int(df['Price'].mode()[0])
rating_mode= int(df['Rating'].mode()[0])
print(price_mode)
print(rating_mode)
```

```python
# Putting values by Mode of it
df['Mode_Price']=df['Price'].fillna(price_mode).astype(int)
df['Mode_Rating']=df['Rating'].fillna(rating_mode).astype(int)
```

```python
# Lets assume price over rating
x_mode=df['Mode_Rating']
y_mode=df['Mode_Price']
```

```python
# By defining B1 and B0
x_mode_mean=x_mode.mean()
y_mode_mean=y_mode.mean()
print(x_mode_mean)
print(y_mode_mean)
```

```python
B1=((x_mode-x_mode_mean)*(y_mode-y_mode_mean)).sum()/((x_mode-x_mode_mean)**2).sum()
B0=y_mode_mean-B1*x_mode_mean
print(B1)
print(B0)
```

```python
y_mode_predicted=B0 + B1*x
y_mode_predicted
```

```python
mode_error=y-y_mode_predicted
mode_error
```
